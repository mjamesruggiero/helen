## data quality check
every number is only as good as categorization; we should shrink the `Uncategorized` tail

```python
out = df[df["is_outflow"]]
total_outflow = -out["amount"].sum()
uncategorized = out[out["category"]=="Uncategorized"]
uncategorized_outflow = -uncategorized["amount"].sum()
percentage = (uncategorized_outflow / total_outflow) * 100
print(f"total_outflow= {total_outflow:.2f} and uncategorized = {uncategorized_outflow:.2f}; percentage: {percentage:.2f}%")
```
